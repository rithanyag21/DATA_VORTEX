```python
import pandas as pd
import numpy as np

def convert_timestamp(value):
    if pd.isna(value):
        return pd.NaT

    value = str(value).strip()

    # Unix timestamp in seconds
    if value.isdigit() and len(value) == 10:
        try:
            return pd.to_datetime(int(value), unit="s")
        except:
            return pd.NaT

    # Date/time formats such as:
    # 25-09-2024
    # 2025-04-13T20:12:18
    try:
        return pd.to_datetime(value, dayfirst=True)
    except:
        return pd.NaT
```

```python
posts_clean["timestamp"] = posts_clean["timestamp"].apply(convert_timestamp)

print(posts_clean["timestamp"].head(20))
```
